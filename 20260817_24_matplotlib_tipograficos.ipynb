```python
x = np.linspace(0, 5, 11)
y1 = x ** 2
y2 = x ** 3

frutas = ['Manzanas', 'Plátanos', 'Peras']
porcentajes = [30, 50, 20]

categorias = ['A', 'B', 'C', 'D']
valores = [40, 70, 30, 90]
plt.figure(figsize=(12,7))
plt.subplot(2, 2, 1)
plt.plot(x, y1, 'bo-')
plt.title('Función Cuadrática')

plt.subplot(2, 2, 2)
plt.plot(x, y2, 'go-')
plt.title('Función Cúbica')

plt.subplot(2, 2, 3)
plt.pie(porcentajes, labels=frutas, autopct='%1.1f%%')
plt.title('Gráfico de pastel')

plt.subplot(2, 2, 4)
plt.bar(categorias, valores, width=0.5, label=categorias,
        color=['red', 'black', 'blue', 'green'])
plt.xlabel('Categoría')
plt.ylabel('Frecuencia')
plt.title('Gráfico de barras verticales')

plt.tight_layout(pad=0.01, h_pad=5, w_pad=10)
plt.show()
```

```python
import numpy as np
from matplotlib import pyplot as plt
from matplotlib.figure import Figure

x = np.linspace(0, 5, 11)
y1 = x ** 2
y2 = x ** 3

fig1 = plt.figure()

ax = fig1.add_subplot()
ax.plot(x, y1, 'ro--')
ax.set_title('Función cuadrática')

plt.show()
```

```python
fig2 = plt.figure()

ax = fig2.add_subplot()
ax.plot(x, y2, 'bo--')
ax.set_title('Función cubica')
```

```python
def grafico_cuadratico(ax, x, y):
    ax.plot(x, y, 'ro--')
    ax.set_title('Función cuadrática')
    ax.grid(True)


def grafico_cubico(ax, x, y):
    ax.plot(x, y, 'bo-')
    ax.set_title('Función cúbica')
    ax.grid(True)


dashboard, axes = plt.subplots(1, 2, figsize=(12, 5))

grafico_cuadratico(axes[0], x, y1)
grafico_cubico(axes[1], x, y2)

dashboard.suptitle('Dashboard')
plt.tight_layout()
plt.show()
```
