```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
import plotly.graph_objects as go
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset("iris")
df.head(10)
```

```python
df["species"].value_counts()
```

```python
df.info()
```

```python
### 1. Gráfico de dispersión básico
#Construye un gráfico de dispersión que muestre la relación entre `sepal_length` y `sepal_width`. Diferencia 
# cada observación según la especie mediante el parámetro `color`. Agrega un título descriptivo y etiquetas claras para ambos ejes.
#Conceptos a practicar: px.

fig = px.scatter(
    df,
    x = "sepal_length",
    y = "sepal_width",
    color="species",
    title="Longitud vs ancho de sepalo por especies",
    labels= {
        "sepal_length":"Longitud del sepalo (cm)",
        "sepal_width": "ancho del sepalo (cm)",
        "species":"Especies"

    }
)
fig.show()
```

```python
#2. Histograma comparativo por especie
#Representa la distribución de `petal_length` mediante un histograma. Usa 20 bins, 
# diferencia las especies con `color`, configura `barmode="overlay"` y aplica una opacidad que permita comparar las distribuciones superpuestas.
fig = px.histogram (
    data_frame=df,
    x = "petal_length",
    color = "species",
    nbins=20,
    barmode="overlay",
    opacity= 0.65,
    title="Distribucion del largo del petalo por especie",
    labels={
        "petal_lenght":"Longitud del petalo (cm)",
        "species":"Especies"
    }
)
fig.show()
```

```python
#3. Boxplot de ancho del pétalo
#Crea un boxplot de `petal_width` para cada especie. Muestra también los puntos individuales de las observaciones. 
# Identifica visualmente qué especie presenta los valores más altos y cuál tiene menor dispersión.
fig = px.box(
    data_frame =df ,
    x= "species",
    y ="petal_width",
    color="species",
    points = "all",
    title= "Ancho del petalo por especie",
    labels={
        "species":"especies",
        "petal_width":"ancho del petalo en cm"
    }


)
fig.show()
```

```python
fig = px.violin(
    data_frame=df,
    x="species",
    y="sepal_length",
    color="species",
    points="all",
    box=True,
    title="Distribucion del largo del sépalo por especie",
    labels={
        "species": "Especie",
        "sepal_length": "Longitud del sépalo (cm)"
    }
)

fig.show()
```

```python
fig = px.scatter_3d(
    data_frame=df,
    x="sepal_length",
    y="sepal_width",
    z="petal_length",
    color="species",
    hover_data=["petal_width"],
    title="Grafico scatter 3D"
)
fig.show()
```
