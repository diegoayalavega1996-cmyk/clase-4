```python
import pandas as pd
df = pd.read_csv("C:/Users/diego/Documents/primer_python/01-python-intro/data/iris_dataset.csv")
df.head(5)
```

```python
#9. Histograma: longitud del sépalo
# Crea un histograma de sepal_length_cm con 10 bins, borde visible y etiquetas. Describe dónde se concentra la mayor
# parte de las observaciones.

import matplotlib.pyplot as plt
plt.hist(df["sepal_length_cm"], bins=10, edgecolor="black")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Frecuencia")
plt.title("Histograma de la longitud del sépalo")
plt.show()
```

```python
#10 Histograma y bins Crea dos histogramas de petal_length_cm: uno con 5 bins y otro con 20 bins. Colócalos lado a lado
 #usando subplots. Explica brevemente cómo cambia la interpretación al modificar los bins.

import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].hist(df["petal_length_cm"], bins=5, color= 'orange',edgecolor="black")
ax[0].set_title("Histograma con 5 bins")
ax[0].set_xlabel("Longitud del pétalo (cm)")
ax[0].set_ylabel("Frecuencia")

ax[1].hist(df["petal_length_cm"], bins=20, edgecolor="black")
ax[1].set_title("Histograma con 20 bins")
ax[1].set_xlabel("Longitud del pétalo (cm)")
ax[1].set_ylabel("Frecuencia")

plt.tight_layout()
plt.show()
```

```python
#11 Boxplot por especie
#Construye un boxplot de sepal_length_cm separado por especie. Etiqueta cada caja con el nombre de
#su especie y compara visualmente las medianas.

import matplotlib.pyplot as plt

setosa = df[df["species"] == "setosa"]["sepal_length_cm"]
versicolor = df[df["species"] == "versicolor"]["sepal_length_cm"]
virginica = df[df["species"] == "virginica"]["sepal_length_cm"]

plt.boxplot(
    [setosa, versicolor, virginica],
    label=["setosa", "versicolor", "virginica"]
)

plt.xlabel("Especie")
plt.ylabel("Longitud del sépalo (cm)")
plt.title("Longitud del sépalo por especie")

plt.show()
```

```python
#12 Boxplot de variables numéricas
#Crea un solo boxplot que compare las cuatro variables numéricas del dataset. Usa etiquetas cortas y
#legibles para cada caja. ¿Qué problema de escala observas?

import matplotlib.pyplot as plt

datos = [
    df["sepal_length_cm"],
    df["sepal_width_cm"],
    df["petal_length_cm"],
    df["petal_width_cm"]
]

plt.boxplot(
    datos,
    label=["Sepal L.", "Sepal W.", "Petal L.", "Petal W."], patch_artist=True
)
plt.xlabel("Grupo")
plt.ylabel("Medida (cm)")
plt.title("Boxplot de variables numéricas")

plt.show()
```

```python
#Subplots 1x2
#Crea una figura con dos subplots: a la izquierda un scatter de sepal_length_cm vs. sepal_width_cm; a
#la derecha un histograma de sepal_length_cm. Agrega un título independiente a cada panel
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].scatter(df["sepal_length_cm"], df["sepal_width_cm"])
ax[0].set_xlabel("Longitud del sépalo (cm)")
ax[0].set_ylabel("Ancho del sépalo (cm)")
ax[0].set_title("Longitud vs. ancho del sépalo")

ax[1].hist(df["sepal_length_cm"], bins=10, color = 'red',edgecolor="black")
ax[1].set_xlabel("Longitud del sépalo (cm)")
ax[1].set_ylabel("Frecuencia")
ax[1].set_title("Histograma de longitud del sépalo")

plt.tight_layout()
plt.show()
```
