```python
def extract_text(ruta: str) -> list[str]:
    mi_lista = []

    with open(ruta, "r") as f:
        contenido = f.read()
        lineas = contenido.split("\n")

        for linea in lineas:
            if "registro" in linea.lower():
                continue
            if "ID" in linea:
                continue

            if not linea:
                continue

            if "instrucciones" in linea.lower():
                break

            mi_lista.append(linea)

   
    return mi_lista
    print(mi_lista)

contenido= extract_text(
        "./data/ventas_polos_aqp.txt"
        
)
print(contenido)
```

```python
class Venta:

    def __init__(
        self,
        codigo: str,
        producto: str,
        talla: str,
        color: str,
        cliente: str,
        ciudad: str,
        cantidad: int,
        precio: float,
        metodo_pago: str,
        estado: str
    ) -> None:

        self.codigo = codigo
        self.producto = producto
        self.talla = talla
        self.color = color
        self.cliente = cliente
        self.ciudad = ciudad
        self.cantidad = cantidad
        self.precio = precio
        self.metodo_pago = metodo_pago
        self.estado = estado

    def total(self):
        return self.cantidad * self.precio

class GestorVentas:

    def __init__(self,registros: list[str]) -> None:

        self.ventas: list[Venta] = []

        for registro in registros:

            info = [item.strip() for item in registro.split(",")]

            venta = Venta(
                codigo=info[0],
                producto=info[1],
                talla=info[2],
                color=info[3],
                cliente=info[4],
                ciudad=info[5],
                cantidad=int(info[6]),
                precio=float(info[7]),
                metodo_pago=info[8],
                estado=info[9]
            )

            self.ventas.append(venta)

    def contar_estados(self):

        entregadas = 0
        pendientes = 0
        canceladas = 0

        for venta in self.ventas:

            if venta.estado == "entregado":
                entregadas += 1

            elif venta.estado == "pendiente":
                pendientes += 1

            elif venta.estado == "cancelado":
                canceladas += 1

        print("Entregadas:", entregadas)
        print("Pendientes:", pendientes)
        print("Canceladas:", canceladas)

    def ingreso_total_entregadas(self):

        total = 0

        for venta in self.ventas:

            if venta.estado == "entregado":
               total += venta.total()

        print(f"Ingreso total de las ventas entregadas:", total ,"soles")

    def producto_mas_vendido(self):

        productos = {}

        for venta in self.ventas:

            if venta.producto in productos:
               productos[venta.producto] += venta.cantidad
            else:
               productos[venta.producto] = venta.cantidad
    

        mayor_producto = ""
        mayor_cantidad = 0

        for producto, cantidad in productos.items():

            if cantidad > mayor_cantidad:
                mayor_cantidad = cantidad
                mayor_producto = producto

        print("Producto más vendido:", mayor_producto)
        print("Cantidad vendida:", mayor_cantidad)

    def ventas_arequipa(self):

        contador = 0

        for venta in self.ventas:

            if venta.ciudad == "Arequipa":
              contador += 1

        print("Ventas realizadas en Arequipa:", contador)

    def ventas_por_ciudad(self, ciudad: str):

       contador1 = 0

       for venta in self.ventas:

          if venta.ciudad.lower() == ciudad.lower():
             contador1 += 1

       print(f"Ventas realizadas en {ciudad}: {contador1}")


    def metodo_pago_mas_usado(self):

       metodos = {}

       for venta in self.ventas:

           if venta.metodo_pago in metodos:
              metodos[venta.metodo_pago] += 1
           else:
              metodos[venta.metodo_pago] = 1

       mayor_metodo = ""
       mayor_cantidad = 0

       for metodo, cantidad in metodos.items():

           if cantidad > mayor_cantidad:
              mayor_cantidad = cantidad
              mayor_metodo = metodo

       print("Método de pago más usado:", mayor_metodo)
       print("Cantidad de ventas:", mayor_cantidad)

    def resumen_productos(self):

       resumen = {}

       for venta in self.ventas:

          if venta.producto in resumen:

              resumen[venta.producto]["unidades"] += venta.cantidad
              resumen[venta.producto]["ingreso"] += venta.total()

          else:

              resumen[venta.producto] = {
                "unidades": venta.cantidad,
                "ingreso": venta.total()
            }

       print("RESUMEN POR PRODUCTO\n")

       for producto, datos in resumen.items():

          print("Producto:", producto)
          print("Unidades vendidas:", datos["unidades"])
          print("Ingreso generado:", datos["ingreso"])
          print("-----------------------------")

    def export_ventas(self, ruta: str) -> None:

        with open(ruta, "w", encoding="utf-8") as f:

             f.write("codigo,producto,talla,color,cliente,ciudad,cantidad,precio,metodo_pago,estado,total\n")

             for venta in self.ventas:

                f.write(
                f"{venta.codigo},"
                f"{venta.producto},"
                f"{venta.talla},"
                f"{venta.color},"
                f"{venta.cliente},"
                f"{venta.ciudad},"
                f"{venta.cantidad},"
                f"{venta.precio},"
                f"{venta.metodo_pago},"
                f"{venta.estado},"
                f"{venta.total()}\n"
            )
```

```python
# 4. Calcula el total vendido por cada registro: cantidad * precio_unitario.
gestor = GestorVentas(contenido)

for venta in gestor.ventas:
    print(
        venta.codigo,
        "-",
        venta.producto,
        "- Total:",
        venta.total()
    )

```

```python
#5. Muestra cuántas ventas fueron entregadas, pendientes y canceladas.
gestor.contar_estados()
```

```python
#6. Calcula el ingreso total solo de las ventas entregadas.
gestor.ingreso_total_entregadas()
```

```python
#7. Muestra cuál fue el producto más vendido por cantidad.
gestor.producto_mas_vendido()
```

```python
#8. Muestra cuántas ventas se hicieron en Arequipa.
gestor.ventas_arequipa()
gestor.ventas_por_ciudad("Arequipa")
gestor.ventas_por_ciudad("Lima")
```

```python
#9. Muestra el método de pago más usado.
gestor.metodo_pago_mas_usado()
```

```python
# 10. Reto extra: crea un resumen por producto con unidades vendidas e ingreso generado.
gestor.resumen_productos()
```
