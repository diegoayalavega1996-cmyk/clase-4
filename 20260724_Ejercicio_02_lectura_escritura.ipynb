```python
def extract_text(ruta: str) -> list[str]:
    mi_lista = []

    with open(ruta, "r") as f:
        contenido = f.read()
        lineas = contenido.split("\n")

        for linea in lineas:
            if "registro" in linea.lower():
                continue
            if "ID" in linea:
                continue

            if not linea:
                continue

            if "instrucciones" in linea.lower():
                break

            mi_lista.append(linea)

   
    return mi_lista
    print(mi_lista)

contenido= extract_text(
        "./data/ventas_polos_aqp.txt"
)
```

```python
class Venta:

    def __init__(
        self,
        codigo: str,
        producto: str,
        talla: str,
        color: str,
        cliente: str,
        ciudad: str,
        cantidad: int,
        precio: float,
        metodo_pago: str,
        estado: str
    ) -> None:

        self.codigo = codigo
        self.producto = producto
        self.talla = talla
        self.color = color
        self.cliente = cliente
        self.ciudad = ciudad
        self.cantidad = cantidad
        self.precio = precio
        self.metodo_pago = metodo_pago
        self.estado = estado

    def total(self):
        return self.cantidad * self.precio

class GestorVentas:

    def __init__(self,registros: list[str]) -> None:

        self.ventas: list[Venta] = []

        for registro in registros:

            info = [item.strip() for item in registro.split(",")]

            venta = Venta(
                codigo=info[0],
                producto=info[1],
                talla=info[2],
                color=info[3],
                cliente=info[4],
                ciudad=info[5],
                cantidad=int(info[6]),
                precio=float(info[7]),
                metodo_pago=info[8],
                estado=info[9]
            )

            self.ventas.append(venta)
```

```python
gestor = GestorVentas(contenido)
print(gestor.ventas[0].codigo)
print(gestor.ventas[0].producto)
print(gestor.ventas[0].cliente)
print(gestor.ventas[0].total())
```
