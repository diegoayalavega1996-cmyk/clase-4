```python
from matplotlib import pyplot as plt
subconjunto = df.loc[0: 29, :]
plt.figure(figsize=(8,4))
plt.grid(True)
plt.plot(subconjunto.index, subconjunto["sepal_length_cm"], marker='o', linestyle= '--')
plt.title('longitud del sepalo en cm - 30 samples')
plt.xlabel('Indic de observaciones')
plt.ylabel('Longitud cm')
plt.show()
```

```python
promedio_sepal_lenght = df.groupby("species")["sepal_length_cm"].mean()
plt.bar(promedio_sepal_lenght.index,promedio_sepal_lenght.values)
plt.title('longitud promedio especies')
plt.xlabel('especies')
plt.ylabel('promedio cm')
plt.show()
```

```python
counts= df["species"].value_counts()
counts
plt.figure(figsize=(6,6))
plt.pie(counts.values, labels= counts.index, autopct='%1.1f%%' )
plt.title('longitud promedio especies')
plt.show()
```

```python
# Visualmente parece existir una relación clara entre ambas variables.

plt.figure(figsize=(7, 5))
plt.scatter(df["sepal_length_cm"], df["sepal_width_cm"])
plt.title("Longitud vs ancho del sépalo")
plt.xlabel("Longitud del sépalo (cm)")
plt.ylabel("Ancho del sépalo (cm)")
plt.show()
```

```python
plt.figure(figsize=(7, 5))

for species, group in df.groupby("species"):
    plt.scatter(
        group["petal_length_cm"],
        group["petal_width_cm"],
        label=species,
    )

plt.title("Longitud vs ancho del pétalo por especie")
plt.xlabel("Longitud del pétalo (cm)")
plt.ylabel("Ancho del pétalo (cm)")
plt.legend(title="Especie")
plt.show()
```

```python
# 15. Leyendas y etiquetas
# Scatter de petal_length_cm vs petal_width_cm por especie

fig, ax = plt.subplots(figsize=(7, 5))

for species, group in df.groupby("species"):
    plt.scatter(
        group["petal_length_cm"],
        group["petal_width_cm"],
        label=species
    )

ax.set_title("Relación entre largo y ancho del pétalo por especie")
plt.xlabel("Largo del pétalo (cm)")
plt.ylabel("Ancho del pétalo (cm)")
plt.legend(title="Especie")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
mean_sepal_width = df.groupby("species")["sepal_width_cm"].mean()
mean_sepal_width.head()
plt.style.use("ggplot")
```

```python
# 16. Estilos
# Promedio de sepal_width_cm por especie

mean_sepal_width = df.groupby("species")["sepal_width_cm"].mean()

# Aplicar temporalmente el estilo ggplot
plt.style.use("ggplot")

fix, ax = plt.subplots(figsize=(7, 5))

ax.bar(mean_sepal_width.index, mean_sepal_width.values)

ax.set_title("Ancho promedio del sépalo por especie")
ax.set_xlabel("Especie")
ax.set_ylabel("Promedio (cm)")

plt.tight_layout()
plt.show()

# Restaurar el estilo original
plt.style.use("default")
```

```python
# 17. Figure y Axes
# Enfoque orientado a objetos

mean_petal_width = df.groupby("species")["petal_width_cm"].mean()

fig, ax = plt.subplots(figsize=(7, 5))

ax.bar(mean_petal_width.index, mean_petal_width.values)

ax.set_title("Ancho promedio del pétalo por especie")
ax.set_xlabel("Especie")
ax.set_ylabel("Promedio (cm)")

plt.tight_layout()
plt.show()
```
