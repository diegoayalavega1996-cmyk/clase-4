```sql
-- Pregunta 01: Muestra nombre y precio de todos los productos, del mas caro al mas barato
SELECT nombre, precio
FROM tienda.productos
ORDER BY precio DESC;
```

```sql
-- Pregunta 02: Lista los clientes de España o Mexico cuyo nombre empeice por M
SELECT *
FROM clientes
WHERE pais IN ('MX', 'ES') AND nombre LIKE 'M%';
```

```sql
-- Pregunta 03: Obten los productos con precio entre 20 y 60  que tengan stock disponible (>0)
SELECT *
FROM productos
WHERE precio BETWEEN 20 AND 60 AND stock > 0;
```

```sql
-- Pregunta 04: Muestra los clientes sin telefono, sustituyendo NULL por 'sin datos'
SELECT nombre, 
COALESCE(telefono, 'sin datos') AS telefono_actualizado
FROM clientes;
```

```sql
SELECT *
FROM tienda.productos
```

```sql
-- Pregunta 05: Lista los 10 productos mas caros con su precio IVA(X 1.21)usando un alias
SELECT nombre, precio * 1.21 AS precio_iva
FROM tienda.productos
ORDER BY precio DESC
LIMIT 10;
```

```sql
-- Pregunta 06: Clasifica los productos en "barato","medio" o "caro" con CASE:
SELECT nombre, precio,
    CASE 
        WHEN precio < 20 THEN 'barato'
        WHEN precio BETWEEN 20 AND 100 THEN 'medio'
        ELSE 'caro'
    END AS clasificacion
FROM productos;
```
