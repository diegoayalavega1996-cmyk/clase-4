```python
import pandas as pd

# Crea una Series con las temperaturas [21, 24, 19, 27, 30] e índice de lunes a viernes.
temperaturas = pd.Series(
    [21, 24, 19, 27, 30],
    index=["Lunes", "Martes", "Miércoles", "Jueves", "Viernes"]
)

print(temperaturas)

# Muestra la media, el máximo y el día que corresponde al máximo (usa .idxmax()).
print("=" * 100)
print("Media: ", temperaturas.mean())
print("Máximo: ", temperaturas.max())
print("Día máximo: ", temperaturas.idxmax())
print("Día máximo índice: ", temperaturas.argmax())
print("=" * 100)

# Crea un DataFrame con 5 productos: nombre, categoría, stock y precio (inventados).
df = pd.DataFrame(
    {
        "nombre": ["Laptop", "Monitor", "Teclado", "Mouse", "Fuente de poder"],
        "categoria": ["Tecnología", "Tecnología", "Tecnología", "Tecnología", "Tecnología"],
        "stock": [3, 7, 12, 1, 0],
        "precio": [1200.0, 320.0, 95.0, 12.23, 90.8]
    }
)

df.head()

# Añade una columna valor_stock = stock × precio.
df["valor_stock"] = df["stock"] * df["precio"]

df.head()

# Imprime la forma del DataFrame y la lista de nombres de columna.
print(df.shape)

print(df.columns.to_list())
```
