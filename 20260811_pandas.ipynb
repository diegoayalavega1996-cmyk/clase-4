```python
s = pd.Series([120, 85, 200, 45],
index=["ene", "feb", "mar", "abr"],
name="ventas")
print(s["mar"]) # acceso por etiqueta
print("***"*100)
print(s * 1.21) # vectorizado:
```

```python
# a) Desde un diccionario de listas (la más habitual)
df = pd.DataFrame ({
      "producto" : ["Laptop", "Monitor" , "Teclado" ],
      "unidades" : [3, 7, 12],
      "precio": [1200.0, 320.0, 95.0],
})

print(df)
```

```python
# b) Desde una lista de diccionarios (típico de una API JSON)
df2 = pd. DataFrame ([{"producto" : "Laptop", "unidades" : 3},
{"producto" : "Monitor" , "unidades" : 7}])
print(df2)
```

```python
# c) Desde una lista de listas, indicando las columnas
df3 = pd. DataFrame ([["Laptop", 3], ["Monitor" , 7]],
columns=[ "producto" , "unidades" ])
print(df3)
```

```python
# d) Desde un array de NumPy
df4 = pd. DataFrame (np.random.rand(3, 2), columns=[ "x", "y"])
print(df4)
```

```python
df = pd.DataFrame ({
      "producto" : ["Laptop", "Monitor" , "Teclado" ],
      "unidades" : [3, 7, 12],
      "precio": [1200.0, 320.0, 95.0],
})
print(df)
print("*"*100)
df["importe" ] = df["unidades" ] * df["precio"]
print(df.shape)
print("*"*100)
print(df.columns.tolist())
print("*"*100)
print(df)
```
