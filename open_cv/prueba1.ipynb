```python
import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        print("No se pudo leer la cámara")
        break

    cv2.imshow("camara", frame)

    if cv2.waitKey(1) == 27:
        break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2  # libreria de opencv

cap = cv2.VideoCapture("http://192.168.1.52:4747/video")  # busca las camaras disponible

while True:  # bucle infinito
    ret, frame = cap.read()  # leer la informacion de la camara

    cv2.imshow('camara', frame)  # mostramos la camara

    k = cv2.waitKey(1)  # espera 1 seg la llave
    if k == 27:  # si alguien presiona 'esc'
        break  # romper el bucle infinito

cap.release()  # apagar la camara
cv2.destroyAllWindows()  # cierra todas las ventanas creadas
```

```python
import cv2  # libreria de opencv

cap = cv2.VideoCapture("cristo.MP4")  # busca las camaras disponible

while True:  # bucle infinito
    ret, frame = cap.read()  # leer la informacion de la camara
    if not ret:
        break
    cv2.imshow('camara', frame)  # mostramos la camara


    k = cv2.waitKey(1)  # espera 1 seg la llave
    if k == 27:  # si alguien presiona 'esc'
        break  # romper el bucle infinito

    if k == ord('s'):
        cv2.imwrite('foto.jpg', frame)

cap.release()  # apagar la camara
cv2.destroyAllWindows()  # cierra todas las ventanas creadas
```

```python
import cv2

cap = cv2.VideoCapture("cristo.MP4")

# Comprobar si el video pudo abrirse
if not cap.isOpened():
    print("Error: no se pudo abrir el video")
else:
    while True:
        ret, frame = cap.read()

        # Si ya no hay más frames, termina el bucle
        if not ret:
            print("Fin del video")
            break

        cv2.imshow("camara", frame)

        # ESC para cerrar
        k = cv2.waitKey(1)

        if k == 27:
            break

cap.release()
cv2.destroyAllWindows()
```

```python
import cv2  # libreria de opencv
import time

cap = cv2.VideoCapture("cristo1.MP4")  # busca las camaras disponible
cont = 0
while True:  # bucle infinito
    ret, frame = cap.read()  # leer la informacion de la camara
    if not ret:
        break
    cv2.imshow('camara', frame)  # mostramos la camara


    k = cv2.waitKey(1)  # espera 1 seg la llave
    if k == 27:  # si alguien presiona 'esc'
        break  # romper el bucle infinito

    if cont < 10:
        cv2.imwrite('foto' + str(cont) + '.jpg', frame)
        cont = cont + 1
        time.sleep(0.1)

cap.release()  # apagar la camara
cv2.destroyAllWindows()  # cierra todas las ventanas creadas
```
